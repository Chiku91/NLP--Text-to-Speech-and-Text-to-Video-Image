```python
import pdfplumber
import pytesseract
from PIL import Image
import pyttsx3
import os

# Set the path to your Tesseract executable (Make sure this path is correct)
# On Windows, the path might be something like:
# "C:\\Program Files\\Tesseract-OCR\\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"  # Update this path

# Function to extract text from the PDF using pdfplumber or OCR if text is not extractable
def extract_text_from_pdf(pdf_path):
    try:
        text = ''
        with pdfplumber.open(pdf_path) as pdf:
            # Loop through all pages and extract text
            for page_num, page in enumerate(pdf.pages):
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"  # Add a newline after each page's text
                else:
                    print(f"Warning: No text extracted from page {page_num + 1}, using OCR...")
                    # Use OCR on the page if no text is found
                    page_image = page.to_image()
                    page_text = pytesseract.image_to_string(page_image.original)
                    text += page_text + "\n"  # Add the OCR-extracted text
        return text
    except FileNotFoundError:
        print(f"Error: The PDF file at {pdf_path} was not found.")
        return ""
    except Exception as e:
        print(f"Error extracting text from PDF: {e}")
        return ""

# Function to convert text to speech and speak it
def text_to_speech(text):
    try:
        # Initialize pyttsx3 TTS engine
        engine = pyttsx3.init()
        engine.setProperty('rate', 150)  # Speed of speech (words per minute)
        engine.setProperty('volume', 1)  # Volume (0.0 to 1.0)

        # Speak the extracted text aloud
        engine.say(text)
        engine.runAndWait()

        print("Text-to-speech conversion completed successfully.")
    except Exception as e:
        print(f"Error during text-to-speech conversion: {e}")

# Main function to execute the process
def main(pdf_path):
    # Extract text from the PDF
    text = extract_text_from_pdf(pdf_path)
    if text:
        print("Text extracted successfully, now converting to speech...")
        # Convert the extracted text to speech and speak it aloud
        text_to_speech(text)
    else:
        print("No text found in the PDF.")

# File path
pdf_path = r"C:\Users\ASUS\Downloads\Graph Theory (Part 03) (Lec 09) _ Class Notes.pdf"

# Check if the PDF file exists before proceeding
if not os.path.exists(pdf_path):
    print(f"Error: The PDF file at {pdf_path} does not exist.")
else:
    # Execute the main function
    main(pdf_path)
```
